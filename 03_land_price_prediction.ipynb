# Land Price Prediction

**Goal:** given the size of a plot of land (in square feet), predict its price.

**What you will learn:**
- The difference between regression and classification (a number vs a category).
- How to visualise a relationship between two columns.
- How to train and use a Linear Regression model.
- Why a straight line is `y = m*x + b`.

**Problem type:** regression (the answer is a number).

**Dataset:** `data/landpriceprediction.csv` - 10 plots with a `land` size and its `price`.

## Import the libraries

```python
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.linear_model import LinearRegression
```

## Load the dataset

The data lives in the `data/` folder, one level up from this notebook. The helper below resolves that relative path whether you launch the notebook from the project root or from inside `notebooks/`.

```python
def find_data(filename):
    for base in [".", "data", "../data", "../../data"]:
        path = os.path.join(base, filename)
        if os.path.exists(path):
            return path
    raise FileNotFoundError(
        f"Could not find {filename}. Keep the data/ folder next to the notebooks/ folder."
    )


DATA_PATH = find_data("landpriceprediction.csv")
df = pd.read_csv(DATA_PATH)

print("Loaded:", DATA_PATH)
print("Rows, columns:", df.shape)
df
```

## Explore the data

```python
df.describe()
```

## Visualise the relationship

Always plot first. Does the price seem to grow with the land size?

```python
plt.figure(figsize=(7, 5))
plt.scatter(df["land"], df["price"], color="green", marker="*")
plt.xlabel("Land (square feet)")
plt.ylabel("Price")
plt.title("Land size vs price")
plt.show()
```

## Choose features and target

- **X** = the input (`land`).
- **y** = the answer we want to predict (`price`).

`X` must be a 2D table, so we keep it as a DataFrame with one column.

```python
X = df[["land"]]
y = df["price"]

print("X shape:", X.shape)
print("y shape:", y.shape)
```

## Train a Linear Regression

Linear Regression fits the straight line that is closest to all the points.

```python
model = LinearRegression()
model.fit(X, y)
```

## Predict and draw the line

```python
land_to_predict = 1500
predicted_price = model.predict(pd.DataFrame([[land_to_predict]], columns=["land"]))
print(f"Predicted price for {land_to_predict} sq ft: {predicted_price[0]:.2f}")

plt.figure(figsize=(7, 5))
plt.scatter(df["land"], df["price"], color="green", marker="*", label="data")
plt.plot(df["land"], model.predict(X), color="red", label="model line")
plt.xlabel("Land (square feet)")
plt.ylabel("Price")
plt.title("Linear Regression fit")
plt.legend()
plt.show()
```

## Check the maths

The model has learned a slope (`coef_`) and an intercept (`intercept_`). Together they define `y = m*x + b`.

```python
m = model.coef_[0]
b = model.intercept_

print(f"slope (m)     = {m:.4f}")
print(f"intercept (b) = {b:.4f}")
```

```python
manual_price = m * land_to_predict + b
print(f"y = m*x + b  ->  {m:.4f} * {land_to_predict} + {b:.4f} = {manual_price:.2f}")
print(f"model.predict gave: {predicted_price[0]:.2f}")
```

## What you learned

- Regression predicts a number; classification predicts a category.
- `LinearRegression` learns a slope and an intercept from the data.
- `predict()` uses that line to make new predictions, and it matches `y = m*x + b`.
